### Streaming and Batch 

### Streaming 

Most models can stream their output content while it is generating response. by dispalying output progressively, streaming signiificantly improves user experience, particular for longer responses. calling stream() returns an iterator that yields output chunks as they are produced you can use a loop to process each chunk in real time

```python
model.invoke("write me a 200 words paragraph on ai/ml")
```

```python
for chunk in model.stream("write me a 200 words paragraph on ai/ml"):
  print(chunk.text, end="", flush=True)
```

### Batch 

Batching a collection of independent requests to a model can significantly improve performance and reduce cost, as the processing can be done in parallel.

```python
responses = model.batch([
    "what is transformers in ai/ml?",
    "what is rag?",
    "what is vectorless rag?"],

    config={
        'max_concurrency': 5, #limit how many parallel calls can be made 
    }

)

for response in responses:
    print(response)
```
